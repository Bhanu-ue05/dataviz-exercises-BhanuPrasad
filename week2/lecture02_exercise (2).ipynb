## Setup — Run this cell first


```python

import pandas as pd
import numpy as np

# World Happiness Report 2023
# Load the CSV file from Kaggle

df = pd.read_csv('/kaggle/input/datasets/bhanuprasadspirit/whr2023-csv/WHR2023.csv')

print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print("Column names:", df.columns.tolist())
print(df.head())
```

```python

print(df.columns.tolist())
```

```python

import plotly.express as px
import pandas as pd

# Select the 8 happiest countries
top8 = df.nlargest(8, 'Ladder score').copy()
top8['Group'] = '8 Happiest Countries'

# Select the 8 least happy countries
bottom8 = df.nsmallest(8, 'Ladder score').copy()
bottom8['Group'] = '8 Least Happy Countries'

# Combine both groups
comparison = pd.concat([top8, bottom8])

# Create the bar chart
fig = px.bar(
    comparison,
    x='Country name',
    y='Ladder score',
    color='Group',
    title='Comparison of the 8 Happiest and 8 Least Happy Countries',
    labels={
        'Country name': 'Country',
        'Ladder score': 'Happiness Score'
    },
    barmode='group'
)

fig.update_layout(
    xaxis_tickangle=-45,
    template='plotly_white'
)

fig.show()
```

## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python

import pandas as pd
import plotly.express as px

# 1. Select the 8 happiest countries
top8 = df.nlargest(8, 'Ladder score').copy()
top8['Group'] = '8 Happiest Countries'

# 2. Select the 8 least happy countries
bottom8 = df.nsmallest(8, 'Ladder score').copy()
bottom8['Group'] = '8 Least Happy Countries'

# 3. Combine both groups
comparison = pd.concat([top8, bottom8], ignore_index=True)

# 4. Calculate the happiness gap and global average
global_avg = df['Ladder score'].mean()
highest_score = top8['Ladder score'].max()
lowest_score = bottom8['Ladder score'].min()
gap = highest_score - lowest_score

# 5. Sort countries so the happiest appear at the top
comparison = comparison.sort_values(
    'Ladder score', ascending=True
)

# 6. Create the bar chart with custom colours
fig = px.bar(
    comparison,
    x='Ladder score',
    y='Country name',
    orientation='h',
    color='Group',
    color_discrete_map={
        '8 Happiest Countries': '#2A9D8F',
        '8 Least Happy Countries': '#E76F51'
    },
    text=comparison['Ladder score'].round(2),
    title=(
        f'A {gap:.2f}-Point Happiness Gap Separates '
        f'the Happiest and Least Happy Countries'
    ),
    labels={
        'Ladder score': 'Happiness Score',
        'Country name': 'Country',
        'Group': 'Country Group'
    }
)

# 7. Add the global average reference line
fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='#264653',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

# 8. Improve readability and styling
fig.update_traces(
    textposition='outside',
    cliponaxis=False
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    title_font_size=20,
    xaxis=dict(
        title='Happiness Score (Ladder score)',
        rangemode='tozero',
        zeroline=True,
        zerolinecolor='black',
        showgrid=True,
        gridcolor='lightgray'
    ),
    yaxis_title='Country',
    legend_title_text='Country Group',
    margin=dict(l=130, r=70, t=120, b=60),
    height=700
)

fig.show()
```

```python
# Prepare data for the grouped bar chart
plot_df = df.groupby('Region', as_index=False).agg({
    'Logged GDP per capita': 'mean',
    'Freedom to make life choices': 'mean'
})

# Rename columns for the chart
plot_df = plot_df.rename(columns={
    'Logged GDP per capita': 'Average GDP per Capita',
    'Freedom to make life choices': 'Average Freedom'
})

display(plot_df.head())
```

```python
import plotly.express as px

fig = px.bar(
    plot_df,
    x='Region',
    y=['Average GDP per Capita', 'Average Freedom'],
    barmode='group',
    title='GDP and Freedom Differ Across Regions',
    labels={
        'Region': 'Region',
        'value': 'Average Value',
        'variable': 'Measure'
    },
    color_discrete_sequence=['#3366CC', '#F28E2B']
)

fig.update_layout(
    yaxis=dict(rangemode='tozero'),
    template='plotly_white',
    title_x=0.5
)

fig.show()
```
